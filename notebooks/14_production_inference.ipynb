PRODUCTION ML PIPELINE

Now we stop treating the notebook as the final product.

Create a reproducible pipeline:

Raw Transaction
      ↓
Validation
      ↓
Feature Engineering
      ↓
Model
      ↓
Probability
      ↓
Threshold
      ↓
Fraud / Genuine
      ↓
Logging

We'll create a simple production-style predictor.

```python
# ================================================================
# STAGE 14 — PRODUCTION INFERENCE PIPELINE
# ================================================================

import joblib
import numpy as np
import pandas as pd

print("=" * 70)
print("STAGE 14 — PRODUCTION INFERENCE PIPELINE")
print("=" * 70)

class FraudDetector:

    def __init__(
        self,
        model,
        threshold=0.18
    ):

        self.model = model
        self.threshold = threshold

    def predict(self, X):

        probability = self.model.predict_proba(
            X
        )[:, 1]

        prediction = (
            probability >= self.threshold
        ).astype(int)

        result = X.copy()

        result["fraud_probability"] = probability

        result["prediction"] = prediction

        result["decision"] = np.where(
            prediction == 1,
            "FRAUD_REVIEW",
            "GENUINE"
        )

        return result


# ------------------------------------------------
# Create detector
# ------------------------------------------------

detector = FraudDetector(
    model=best_xgb,
    threshold=0.18
)

# ------------------------------------------------
# Example inference
# ------------------------------------------------

sample = X_val.iloc[:10]

result = detector.predict(sample)

display(result[
    [
        "fraud_probability",
        "prediction",
        "decision"
    ]
])

# ------------------------------------------------
# Save model
# ------------------------------------------------

joblib.dump(
    best_xgb,
    "fraud_xgboost_model.pkl"
)

joblib.dump(
    detector,
    "fraud_detector.pkl"
)

print("\nSaved:")
print("fraud_xgboost_model.pkl")
print("fraud_detector.pkl")
```
